```python
from datetime import datetime, timedelta
import random 
import pandas as pd 

# Parâmetros
datas = pd.date_range(start='2024-01-01', end='2024-12-31', freq='W')  # uma venda por semana
regioes = ['Sudeste', 'Sul', 'Nordeste', 'Centro-Oeste', 'Norte']

df_produtos = pd.read_csv('produtos.csv')

# Gerar vendas
vendas = []

for data in datas:
    for _ in range(30):  # 30 vendas por semana
        id_produto = random.choice(df_produtos['id_produto'])
        regiao = random.choice(regioes)
        quantidade = random.randint(1, 20)
        preco_unitario = round(random.uniform(30, 800), 2)
        vendas.append({
            'data': data,
            'id_produto': id_produto,
            'quantidade_vendida': quantidade,
            'preco_unitario': preco_unitario,
            'regiao': regiao
        })

df_vendas = pd.DataFrame(vendas)
df_vendas.to_csv('vendas.csv', index=False)
print(df_vendas)
```
